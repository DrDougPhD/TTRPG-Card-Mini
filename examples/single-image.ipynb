```python
import pathlib
import string
from PIL import Image, ImageFont, ImageDraw

image_path = pathlib.Path("../data/inputs/gnolls/Gnoll-fenar-cardstock.webp")
image = Image.open(image_path)

display(image)
```

```python
WIDTH_INDEX = 2
true_type_font_file = "../fonts/Open_Sans/OpenSans-Regular.ttf"
text_color = (0, 0, 0)

title = "Gnoll"
i = 0
letter = string.ascii_uppercase[i]
indexed_title = f"{title} {letter}"
image_width = image.size[0]

font_size = 1
font = ImageFont.truetype(true_type_font_file, font_size)
rendered_text_width = font.getbbox(indexed_title)[WIDTH_INDEX]
while rendered_text_width < image_width:
    # iterate until the text size is just larger than the criteria
    print(f"{font_size} size: {rendered_text_width}")
    font_size += 1
    font = ImageFont.truetype(true_type_font_file, font_size)
    rendered_text_width = font.getbbox(indexed_title)[WIDTH_INDEX]

# optionally de-increment to be sure it is less than criteria
font_size -= 1
font = ImageFont.truetype(true_type_font_file, font_size)

print(f"Font size for image: {font_size}")

image_drawer = ImageDraw.Draw(image)
image_drawer.text(
    (0, 0),
    indexed_title,
    text_color,
    font=font,
)
display(image)
```
